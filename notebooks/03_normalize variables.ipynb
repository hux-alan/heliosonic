# Normalizing ACE SWEPAM Variables

Before mapping them to sound, this notebook will investigate different metods for transforming physical spacecraft measurements into normalized values suitable for sonifictoin. 

```python
import numpy as np
import matplotlib.pyplot as plt
from cdflib import CDF

cdf = CDF("../data/ac_h0_swe_20240709_v11.cdf")

def min_max_normalize(data):
    return (data - np.min(data)) / (np.max(data) - np.min(data))
```

```python
density = cdf.varget("Np")
speed = cdf.varget("Vp")
temperature = cdf.varget("Tpr")
n_attrs = cdf.varattsget("Np")
v_attrs = cdf.varattsget("Vp")
tpr_attrs = cdf.varattsget("Tpr")


density_valid = density[
    (density != n_attrs["FILLVAL"]) &
    (density >= n_attrs["VALIDMIN"]) &
    (density <= n_attrs["VALIDMAX"])
]

speed_valid = speed[
    (speed != v_attrs["FILLVAL"]) &
    (speed >= v_attrs["VALIDMIN"]) &
    (speed <= v_attrs["VALIDMAX"])
]

tpr_valid = temperature[
    (temperature != tpr_attrs["FILLVAL"]) &
    (temperature >= tpr_attrs["VALIDMIN"]) &
    (temperature <= tpr_attrs["VALIDMAX"])
]

# Check before normalizing

print("----------------")
print("Tpr")
print("Original samples:", len(temperature))
print("Valid samples:", len(tpr_valid))

print("----------------")
print("Np")
print("Original samples:", len(density))
print("Valid samples:", len(density_valid))

print("----------------")
print("Vp")
print("Original samples:", len(speed))
print("Valid samples:", len(speed_valid))
```
